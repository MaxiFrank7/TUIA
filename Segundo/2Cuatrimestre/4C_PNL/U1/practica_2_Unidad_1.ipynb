## Práctica: extracción, limpieza y consulta de un diccionario de lunfardo

### Ejercicio 1 — Del PDF a una tabla

A partir del PDF provisto, realice un proceso que permita extraer su contenido textual y transformarlo en una tabla de datos.

El objetivo final es obtener un DataFrame con, al menos, las siguientes columnas:

- `termino`
- `definicion`

Para ello, deberá:

1. extraer el texto del PDF;
2. inspeccionar el resultado de la extracción;
3. identificar errores y problemas del texto;
4. construir una función de normalización y estandarización;
5. separar correctamente cada entrada del diccionario;
6. estructurar la información en formato tabular.

La función de normalización debe intentar corregir, en la medida de lo posible:

- errores de mayúsculas y minúsculas;
- espacios sobrantes;
- signos de puntuación mal ubicados;
- tildes ausentes;
- errores frecuentes de tipeo;
- otras irregularidades detectadas en el documento.

No se espera una corrección perfecta, sino una estrategia clara, razonable y reproducible.

#### Preguntas de cierre

- ¿Qué problemas no se resolvieron del todo?
- ¿Qué parte del proceso resultó más compleja?

### Ejercicio 2 — Buscar términos a partir de una frase

A partir de la tabla obtenida en el ejercicio anterior, construya un programa que permita ingresar una palabra, frase o breve descripción y recuperar los **tres términos más próximos** del diccionario.

Para ello, el sistema debe:

1. recibir una consulta del usuario;
2. normalizar esa consulta con la misma función usada en el ejercicio 1;
3. eliminar stopwords;

5. comparar la consulta con las definiciones a partir de las palabras compartidas;
6. devolver los tres términos más cercanos junto con sus definiciones.

La comparación debe realizarse únicamente con herramientas básicas de procesamiento de texto. No deben utilizarse métodos de representación vectorial avanzada ni herramientas de búsqueda más complejas.

## Ejercicio 3 — Representar las definiciones como vectores

En los ejercicios anteriores, la recuperación de términos se realizó mediante una comparación simple entre palabras de la consulta y palabras de las definiciones. Si bien esta estrategia es útil como punto de partida, presenta limitaciones: depende demasiado de coincidencias exactas y puede volverse poco eficiente cuando aumenta la cantidad de definiciones a comparar.

En este ejercicio, se propone reemplazar esa comparación directa por una representación vectorial clásica del texto.

A partir de la tabla generada en el ejercicio 1, y utilizando la columna de definiciones normalizadas, construya distintas representaciones vectoriales de las definiciones. Para ello, trabaje con los siguientes métodos:

- one-hot
- CountVectorizer
- TF-IDF
- Hashing Vectorizer

El objetivo es transformar cada definición en un vector numérico para poder comparar luego la consulta del usuario en un espacio común de representación.

### Consigna

A partir del DataFrame previamente construido:

1. tome la columna de definiciones normalizadas;
2. genere una representación vectorial del corpus usando cada uno de los métodos indicados;
3. describa brevemente qué información conserva cada representación;
4. compare el tamaño y la estructura de las matrices obtenidas;
5. observe qué diferencias aparecen entre una representación basada en presencia, una basada en frecuencia, una ponderada y una basada en hashing.

### Orientaciones

- En el caso de one-hot, puede construir una versión simple donde cada palabra del vocabulario indique presencia o ausencia.
- En CountVectorizer, cada vector debe reflejar la frecuencia de aparición de las palabras en la definición.
- En TF-IDF, cada palabra debe quedar ponderada según su importancia relativa en el corpus.
- En Hashing Vectorizer, la representación debe realizarse sin construir explícitamente un vocabulario.

### Preguntas de cierre

- ¿Qué diferencias conceptuales existen entre presencia, frecuencia y ponderación?
- ¿Qué ventaja tiene TF-IDF respecto de CountVectorizer?
- ¿Qué ventaja práctica ofrece una representación basada en hashing?
- ¿Qué información puede perderse o volverse menos interpretable con Hashing Vectorizer?

La idea es evaluar cómo cambia el comportamiento del buscador al reemplazar la comparación simple por modelos clásicos de representación vectorial.
