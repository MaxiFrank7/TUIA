1. Calcule y grafique las matrices de correlación y covariancia.

```python
msn=pd.read_csv('./datasets/estacion_meteorologica.csv', on_bad_lines='skip', sep=';')
msn
```

```python
msn.info()
```

```python
msn.columns = msn.columns.str.strip()
```

```python
msn.columns.tolist()
```

```python
columnas_convertir=['ColdJunc0','PowerVolt','WS(ave)','WS(inst_m)','Solar_rad','TEMP','Humidity','Rainfall','Bar_press.','Soil_temp']
columnas_numericas=['ColdJunc0','PowerVolt','PowerKind','WS(ave)','WD(ave)','WS(inst_m)','WD(inst_m)','Solar_rad','TEMP','Humidity','Rainfall','Bar_press.','Soil_temp']
```

```python
for columna in columnas_convertir:
    msn[columna]=msn[columna].str.replace(',','.').astype(float)
```

```python
msn.info()
```

```python
sns.heatmap(msn[columnas_numericas].corr(),annot=True,cmap='coolwarm')
plt.title('Matriz correlacion')
plt.figure(figsize=(12, 20))
plt.show()
```

Las de mayor correlacion son humidity;coldjun0 y TEMP;Humidity ambas de -0.96, son inversamente lineal. TEMP;ColdJunc0 es 0.98

Las de menor correlacion son WD(ave);WS(ave) de 0.015 y Rainfall;Soil_temp 0.006.

```python
sns.heatmap(msn[columnas_numericas].cov(),annot=True)
plt.title('Matriz correlacion')
plt.figure(figsize=(12, 20))
plt.show()
```
