```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
msn=pd.read_csv('./datasets/estacion_meteorologica.csv', on_bad_lines='skip', sep=';', decimal=',')
msn
```

```python
msn.info()
```

```python
msn.columns.str.strip()
```

```python
col_num=['ColdJunc0','PowerVolt','PowerKind','WS(ave)','WD(ave)','WS(max)','WS(inst_m)','WD(inst_m)','Solar_rad','TEMP','Humidity','Rainfall','Bar_press.','Soil_temp']
```

```python
sns.heatmap(msn[col_num].corr(),annot=True,cmap='coolwarm')
plt.title('Matriz correlacion')
plt.figure(figsize=(12, 20))
plt.show()
```

Las de mayor correlacion son humidity;coldjun0 y TEMP;Humidity ambas de -0.96, son inversamente lineal. TEMP;ColdJunc0 es 0.98

Las de menor correlacion son WD(ave);WS(ave) de 0.015 y Rainfall;Soil_temp 0.006.
