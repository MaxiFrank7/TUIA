### **Ejercicio N°4**

Teniendo en cuenta la variable `altura_m` que se encuentra en el dataset `pacientes.csv` trabajado en el **Ejercicio N°2**, genere una tabla de frecuencias en la que las observaciones se encuentren segmentadas en subintervalos de 10 cm de amplitud que estén “cerrados por izquierda”, es decir, que tengan la forma **[extremo_inferior, extremo_superior)**.

La tabla de frecuencias generada deberá contener columnas en las que se especifiquen las frecuencias absolutas, relativas y relativas acumuladas correspondientes a cada subintervalo.
¿Qué porcentaje de las personas del dataset tienen una altura **menor a 1.8 m**?


```python
pacientes=pd.read_csv('./datasets/pacientes.csv', sep=';')
pacientes
```

```python
pacientes.info()
```

```python
pacientes['altura_m'].unique().tolist()
```

```python
pacientes['altura_m']=pacientes['altura_m'].str.replace(',','.')
```

```python
pacientes['altura_m']=pacientes['altura_m'].astype('float')
```

```python
sep=np.arange(1.5,2.0,0.1)
pacientes['altura_bins']=pd.cut(pacientes['altura_m'],bins=sep,right=False)
```

```python
pacientes
```

```python
frec_abs=pacientes['altura_bins'].value_counts().sort_index()
frec_rel=pacientes['altura_bins'].value_counts(normalize=True).sort_index()
```

```python
tabla=pd.DataFrame({'Cantidad':frec_abs,
                    'Porcentaje':(frec_rel*100).round(2),
                    'Frecuencia relativa acumulada': frec_rel.cumsum()})
tabla
```

```python
tabla = tabla.rename_axis('Intervalo de alturas')
tabla
```

```python
menos_180=(pacientes['altura_m']<1.80).sum()
print(f'Hay {menos_180} personas que miden menos de 1.80 m que representa un {menos_180*100/len(pacientes)} % del total')
```
