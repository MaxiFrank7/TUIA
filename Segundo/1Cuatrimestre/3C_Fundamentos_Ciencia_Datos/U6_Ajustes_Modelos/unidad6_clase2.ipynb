```python
# Librerías
import pandas as pd
import statsmodels.formula.api as smf
```

```python
# Datos
datos = pd.read_csv(ruta + '/datasets/advertising.csv')
datos.head()
```

## 2. Ajustá un modelo de regresión lineal múltiple para Sales usando los tres predictores: TV, Radio y Newspaper. ¿Cuáles predictores son estadísticamente significativos al 5%?

Inversión en TV y Radio

```python
modelo1 = smf.ols(formula = 'Sales ~ TV + Radio +	Newspaper', data = datos).fit()
print(modelo1.summary())
```

## 3. Reajustá el modelo eliminando el o los predictores no significativos. Usando el modelo reducido, estimá las ventas esperadas para un mercado con una inversión de 150 mil USD en TV, 30 mil USD en radio y 60 mil USD en periódicos.

```python
modelo2 = smf.ols(formula = 'Sales ~ TV + Radio', data = datos).fit() # Ajustamos modelo reducido
print(modelo2.summary())
```

```python
# Predicción de ventas para un mercado
# con una inversión de 150 mil USD en TV, 30 mil USD en radio y 60 mil USD en periódicos.

datos_prediccion = pd.DataFrame({'TV' : [150],
                                 'Radio' : [30],
                                 'Newspaper' : [60]})
datos_prediccion
```

```python
modelo2.predict(datos_prediccion) # 16 mil unidades vendidas
```
