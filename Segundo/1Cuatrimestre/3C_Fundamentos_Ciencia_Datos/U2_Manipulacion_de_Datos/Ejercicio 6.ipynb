```python
import re

# 6.1 Validar URL
def validar_url(url):
    # ^ (inicio), https? (http o https), ://, dominio sin espacios, . extension
    patron = r'^https?://[\w\.-]+\.[a-zA-Z]{2,}$'
    if re.match(patron, url):
        print('URL válida')
    else:
        print('URL no válida')

# 6.2 Validar Gmail
def es_gmail_valido(correo):
    # Letras, números, puntos, etc., seguido estrictamente de @gmail.com
    patron = r'^[a-zA-Z0-9._%+-]+@gmail\.com$'
    return bool(re.match(patron, correo))

# 6.3 Validar Fecha YYYY-MM-DD
def es_fecha_valida(fecha):
    # \d{4} (4 dígitos), 0[1-9]|1[0-2] (meses 01-12), 0[1-9]|[12]\d|3[01] (días 01-31)
    patron = r'^\d{4}-(0[1-9]|1[0-2])-(0[1-9]|[12]\d|3[01])$'
    return bool(re.match(patron, fecha))

# 6.4 Análisis de la canción de Manu Chao
with open('./datasets/Me_gustas_tu-Manu_Chao.txt', 'r', encoding='utf-8') as f:
    letra = f.read().lower()

# Verbo gustar (buscamos "gusta", "gustas", "gustan")
gustar_matches = re.findall(r'\bgusta[sn]?\b', letra)
print(f"Referencias al verbo gustar: {len(gustar_matches)}")

# Verbos en infinitivo (palabras que terminen en ar, er, ir)
# Usamos set() por si queremos saber cuántos verbos ÚNICOS hay
infinitivos = re.findall(r'\b\w+[aei]r\b', letra)
print(f"Total de verbos en infinitivo (apariciones): {len(infinitivos)}")
print(f"Verbos únicos en infinitivo: {len(set(infinitivos))}")

# Cosas que le gustan (Capturamos lo que está entre "me gusta" y la coma o final)
cosas_matches = re.findall(r'me gustan?\s([^,]+)(?:,|\s)', letra)
# Limpiamos espacios en blanco
cosas_que_le_gusta = [cosa.strip() for cosa in cosas_matches if cosa.strip()]
print(f"Cosas que le gustan: {cosas_que_le_gusta}")
```
