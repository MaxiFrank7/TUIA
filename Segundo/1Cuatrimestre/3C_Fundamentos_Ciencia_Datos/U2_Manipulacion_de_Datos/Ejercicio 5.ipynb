### **Ejercicio N°5**

Usando los datos de `listings_ba.csv` de Buenos Aires que se encuentran cargados en el campus realice la imputación de los precios de alquiler faltantes empleando:

1. La media y moda de los datos no faltantes 

2. Una medida de resumen de su elección por barrio y tipo de habitación

3. Los 10 puntos más cercanos geográficamente a cada dato faltante usando las coordenadas.

En cada uno de los casos indicar la cantidad de datos que se usaron para la imputación.

```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
import numpy as np
```

```python
alquileres=pd.read_csv('./datasets/listings_ba.csv')
alquileres
```

```python
alquileres.info()
```

```python
#precio media y moda de todos
df_a=alquileres.copy()
precio_media=df_a['price'].mean()
precio_moda=df_a['price'].mode()[0]
print(f'Media= ${round(precio_media,2)} y moda= ${round(precio_moda,2)}')
```

```python
df_a['price']=df_a['price'].fillna(precio_media)
df_a
```

```python
#medida resumen por barrio y tipo de habitacion
# uso transform para obtener la misma cantidad de rows, sino obtengo muchas menos
df_b=alquileres.copy()
#round(df_b.groupby(['neighbourhood','room_type'])['price'].mean(),2)
precio_media_barrio_room=round(df_b.groupby(['neighbourhood', 'room_type'])['price'].transform('mean'),2)
```

```python
df_b['price']=df_b['price'].fillna(precio_media_barrio_room)
df_b.info()
```

```python
df_b[['id','name','neighbourhood','room_type','price']]
```

```python
# alquileres con precio NA
dfNA=alquileres[alquileres['price'].isna()]
dfNA
```

```python
lista_ids = dfNA['id'].tolist()
```

```python
#NA llenos en df_a
resultadoA=df_a[df_a['id'].isin(lista_ids)]
resultadoA[['id','name','neighbourhood','room_type','price']]
```

```python
#NA llenos en df_b
resultadoB=df_b[df_b['id'].isin(lista_ids)]
resultadoB[['id','name','neighbourhood','room_type','price']]
```
