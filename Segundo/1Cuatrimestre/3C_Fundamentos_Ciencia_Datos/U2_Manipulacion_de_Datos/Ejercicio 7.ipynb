```python
import pandas as pd
```

```python
#Cargamos los datos
df_personas = pd.read_excel('./datasets/conicet_personas_2020.xlsx')
df_sexo = pd.read_excel('./datasets/conicet_ref_sexo.xlsx')
df_grado = pd.read_excel('./datasets/conicet_ref_grado_academico.xlsx')

# Hacemos el cruce (Merge)
df_merged = pd.merge(df_personas, df_sexo, on='sexo_id', how='inner')
df_merged = pd.merge(df_merged, df_grado, left_on='maximo_grado_academico_id', right_on='grado_academico_id', how='inner')

#Generamos la tabla (crosstab es genial para contar frecuencias entre dos categorías)
tabla_conicet = pd.crosstab(df_merged['descripcion'], df_merged['sexo_descripcion'])
print(tabla_conicet)
```
