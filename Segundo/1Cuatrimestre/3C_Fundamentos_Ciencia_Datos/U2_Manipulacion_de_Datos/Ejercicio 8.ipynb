```python
import pandas as pd
```

```python
# Cargamos saltando las primeras 3 filas y arreglando el encoding para los acentos
df_incendios = pd.read_csv('./datasets/incendios-cantidad-causas-provincia_2022.csv', encoding='latin-1', skiprows=3)

# Generamos la tabla con una pivot_table
tabla_incendios = df_incendios.pivot_table(
    values='intencional', 
    index='anio', 
    columns='provincia', 
    aggfunc='sum'
)

print(tabla_incendios)
```
