```python
# Para usar esto en Colab, primero ejecutá en una celda: !pip install jellyfish Levenshtein
import Levenshtein
import jellyfish
import pandas as pd
```

```python

pares = [
    ("Mariana", "Merianna"),
    ("Della Ceca", "Dellacecca"),
    ("Córdoba 2568", "Cordoba 2478")
]

for s1, s2 in pares:
    dist_lev = Levenshtein.distance(s1, s2)
    dist_jaro = jellyfish.jaro_similarity(s1, s2)
    dist_jw = jellyfish.jaro_winkler_similarity(s1, s2)
    
    print(f"'{s1}' vs '{s2}':")
    print(f" - Levenshtein: {dist_lev}")
    print(f" - Jaro: {dist_jaro:.4f}")
    print(f" - Jaro-Winkler: {dist_jw:.4f}\n")
```
